1.CATEGORY: Identifying the Next Winning Cuisine
 Problem

Find which cuisine type has:

High ratings

Strong demand (votes)

But low competition (few restaurants)

Graphs to Use
1. Scatter plot (Average_Rating vs. Total_Votes, color = Cuisine, size = restaurant count)  
2. Bar chart (Cuisine vs. Count of Restaurants)

3. Box plot (Cuisine vs. Avg_Price_Cuisine)
4.  Heatmap of Avg_Rating_Cuisine and Avg_Price_Cuisine
5. Bubble chart combining votes, ratings, and price





```python
# Create a new column 'Restaurant_Count' which contains the count of occurrences of each restaurant name
d['Restaurant_Count'] = d.groupby('Restaurant_Name')['Restaurant_Name'].transform('count')

# Create a scatter plot of 'Average_Rating' versus 'Total_Votes',
# coloring the points by 'Cuisine', and sizing them by 'Restaurant_Count'
plt.figure(figsize=(12, 8))
sns.scatterplot(data=d, x='Total_Votes', y='Average_Rating', hue='Cuisine', size='Restaurant_Count', alpha=0.6)

# Add titles and labels for clarity
plt.title('Average Rating vs Total Votes by Cuisine and Restaurant Count')
plt.xlabel('Total Votes')
plt.ylabel('Average Rating')

# Display the plot
plt.show()
```

```python
# Count the number of restaurants for each cuisine
cuisine_counts = d['Cuisine'].value_counts().reset_index()
cuisine_counts.columns = ['Cuisine', 'Restaurant_Count']

# Create a bar chart
plt.figure(figsize=(15, 7))
sns.barplot(data=cuisine_counts, x='Cuisine', y='Restaurant_Count', palette='viridis') # Changed palette to 'viridis'

# Rotate x-axis labels for better readability
plt.xticks(rotation=90)

# Add titles and labels
plt.title('Count of Restaurants by Cuisine')
plt.xlabel('Cuisine')
plt.ylabel('Number of Restaurants')

# Adjust layout to prevent labels from overlapping
plt.tight_layout()

# Display the plot
plt.show()
```

```python
# Create a box plot of 'Avg_Price_Cuisine' for each cuisine
plt.figure(figsize=(15, 7))
sns.boxplot(data=d, x='Cuisine', y='Avg_Price_Cuisine')

# Rotate x-axis labels for better readability
plt.xticks(rotation=90)

# Add titles and labels
plt.title('Distribution of Average Price per Cuisine')
plt.xlabel('Cuisine')
plt.ylabel('Average Price per Cuisine')

# Adjust layout to prevent labels from overlapping
plt.tight_layout()

# Display the plot
plt.show()
```

```python
# Calculate the average rating and average price for each cuisine
cuisine_avg_metrics = d.groupby('Cuisine')[['Avg_Rating_Cuisine', 'Avg_Price_Cuisine']].mean().reset_index()

# Create a pivot table for the heatmap
heatmap_data = cuisine_avg_metrics.set_index('Cuisine')

# Create a heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(heatmap_data, annot=True, cmap='YlGnBu', fmt=".2f")

# Add titles and labels
plt.title('Heatmap of Average Rating and Average Price per Cuisine')
plt.xlabel('Metric')
plt.ylabel('Cuisine')

# Adjust layout
plt.tight_layout()

# Display the plot
plt.show()
```

```python
# Aggregate data by Cuisine to get average rating, average price, and sum of total votes
cuisine_bubble_data = d.groupby('Cuisine').agg(
    Avg_Rating_Cuisine=('Average_Rating', 'mean'),
    Avg_Price_Cuisine=('Prices', 'mean'),
    Total_Votes=('Total_Votes', 'sum')
).reset_index()

# Create a bubble chart
plt.figure(figsize=(14, 8))
sns.scatterplot(data=cuisine_bubble_data, x='Avg_Price_Cuisine', y='Avg_Rating_Cuisine', size='Total_Votes', hue='Cuisine', alpha=0.7, sizes=(100, 2000))

# Add titles and labels
plt.title('Bubble Chart of Cuisine Metrics (Avg Rating, Avg Price, Total Votes)')
plt.xlabel('Average Price per Cuisine')
plt.ylabel('Average Rating per Cuisine')

# Add legend for bubble size
plt.legend(title='Cuisine', bbox_to_anchor=(1.05, 1), loc='upper left')

# Adjust layout
plt.tight_layout()

# Display the plot
plt.show()
```

Insights to Aim For

1. High-rating
2. High-demand cuisines with low saturation → ideal for market entry.

Example: “Italian and Thai cuisines have top ratings but fewer total restaurants.”

2.MARKET: Pinpointing the Best City / Location
Problem

Find which city or place offers the best potential for growth:

1. High consumer engagement (votes)

2. High average rating

3. Low number of established competitors (Is_Highly_Rated restaurants)

Graphs to Use

1.   Bar chart (City vs. Avg_Rating_City)
2.   Scatter plot (Avg_Rating_City vs. Total_Votes)
3.   Stacked bar chart (City vs. Count of Is_Highly_Rated and total restaurants)
4.   Heatmap or Bubble plot (Place_Name vs. Restaurant_Popularity)
5.   Line/Bar chart comparing Avg_Price_City and Avg_Rating_City



```python
# Calculate the average rating for each city
city_avg_rating = d.groupby('City')['Avg_Rating_City'].mean().reset_index()

# Create a bar chart
plt.figure(figsize=(10, 6))
sns.barplot(data=city_avg_rating, x='City', y='Avg_Rating_City', palette='viridis') # Changed palette to 'viridis' as a suitable violet-to-green option

# Add titles and labels
plt.title('Average Rating by City')
plt.xlabel('City')
plt.ylabel('Average Rating')

# Adjust layout
plt.tight_layout()

# Display the plot
plt.show()
```

```python
# Aggregate data by City to get average rating and sum of total votes
city_metrics = d.groupby('City').agg(
    Avg_Rating_City=('Avg_Rating_City', 'mean'),
    Total_Votes=('Total_Votes', 'sum')
).reset_index()

# Create a scatter plot
plt.figure(figsize=(10, 6))
sns.scatterplot(data=city_metrics, x='Total_Votes', y='Avg_Rating_City', hue='City', s=200) # Increased size of points for better visibility

# Add titles and labels
plt.title('Average Rating vs Total Votes by City')
plt.xlabel('Total Votes')
plt.ylabel('Average Rating')

# Add legend
plt.legend(title='City')

# Adjust layout
plt.tight_layout()

# Display the plot
plt.show()
```

```python
# Aggregate data by City to get the total count and the count of highly-rated restaurants
city_highly_rated = d.groupby('City').agg(
    Total_Restaurants=('Restaurant_Name', 'count'),
    Highly_Rated_Restaurants=('Is_Highly_Rated', lambda x: (x == 1).sum())
).reset_index()

# Create a stacked bar chart
plt.figure(figsize=(12, 7))

# Plot the total restaurants
sns.barplot(data=city_highly_rated, x='City', y='Total_Restaurants', color='darkgreen', label='Total Restaurants') # Changed color to darkgreen

# Plot the highly-rated restaurants on top
sns.barplot(data=city_highly_rated, x='City', y='Highly_Rated_Restaurants', color='maroon', label='Highly Rated Restaurants') # Changed color to maroon

# Add titles and labels
plt.title('Total vs Highly Rated Restaurants by City')
plt.xlabel('City')
plt.ylabel('Number of Restaurants')

# Add legend
plt.legend()

# Adjust layout
plt.tight_layout()

# Display the plot
plt.show()
```

```python
# Calculate the average restaurant popularity for each place name
place_popularity = d.groupby('Place_Name')['Restaurant_Popularity'].mean().reset_index()

# Create a pivot table for the heatmap
heatmap_data_place = place_popularity.set_index('Place_Name')

# Create a heatmap
plt.figure(figsize=(12, 30)) # Increased the height of the figure

# Create a heatmap
sns.heatmap(heatmap_data_place, annot=True, cmap='YlGnBu', fmt=".2f")

# Add titles and labels
plt.title('Heatmap of Average Restaurant Popularity by Place Name')
plt.xlabel('Restaurant Popularity')
plt.ylabel('Place Name')

# Adjust layout
plt.tight_layout()

# Display the plot
plt.show()
```

```python
# Aggregate data by City to get average price and average rating
city_comparison_data = d.groupby('City').agg(
    Avg_Price_City=('Avg_Price_City', 'mean'),
    Avg_Rating_City=('Avg_Rating_City', 'mean')
).reset_index()

# Create a figure and a set of subplots with a secondary y-axis
fig, ax1 = plt.subplots(figsize=(12, 7))
ax2 = ax1.twinx()

# Create a bar chart for Avg_Price_City
sns.barplot(data=city_comparison_data, x='City', y='Avg_Price_City', ax=ax1, color='gold', label='Average Price') # Changed color to gold

# Create a line chart for Avg_Rating_City on the secondary y-axis
sns.lineplot(data=city_comparison_data, x='City', y='Avg_Rating_City', ax=ax2, color='red', marker='o', label='Average Rating')

# Add titles and labels
ax1.set_title('Comparison of Average Price and Average Rating by City')
ax1.set_xlabel('City')
ax1.set_ylabel('Average Price per City')
ax2.set_ylabel('Average Rating per City')

# Add legends
ax1.legend(loc='upper left')
ax2.legend(loc='upper right')

# Adjust layout
fig.tight_layout()

# Display the plot
plt.show()
```

Insights to Aim For

1. “Which city has high demand but fewer top restaurants?”

2. “Which neighborhood shows strong popularity with limited supply?”
